```python
import pandas as pd
import numpy as np




s1 = pd.Series([10, 20, 30, 40], name="MyList")
print("From List:\n", s1, "\n")


s2 = pd.Series({'a': 100, 'b': 200, 'c': 300})
print("From Dictionary:\n", s2, "\n")


s3 = pd.Series(np.random.randn(3), index=['x', 'y', 'z'])
print("From NumPy Array:\n", s3, "\n")

print("=== PANDAS DATAFRAMES ===")

df1 = pd.DataFrame({
    'A': [1, 2, 3],
    'B': ['X', 'Y', 'Z']
})
print("From Dict of Lists:\n")
display(df1)

df2 = pd.DataFrame([
    {'Name': 'Alice', 'Age': 25},
    {'Name': 'Bob', 'Age': 30}
])
print("\nFrom List of Dicts:\n")
display(df2)

df3 = pd.DataFrame(
    np.random.randint(10, 50, size=(3, 3)),
    columns=['Col_1', 'Col_2', 'Col_3'],
    index=['Row_A', 'Row_B', 'Row_C']
)
print("\nFrom 2D NumPy Array:\n")
display(df3)
```
